```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read first 10 rows
df = pd.read_csv("Top Indian Places to Visit.csv").head(10)

# Convert numeric columns
cols = ["Google review rating",
        "Entrance Fee in INR",
        "time needed to visit in hrs"]

for c in cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# Replace "Place Name" with your actual place column if different
x = df["Name"]

# 1. Line Chart
plt.figure(figsize=(6,3))
plt.plot(x, df["Google review rating"], marker="o")
plt.title("Line Chart")
plt.xticks(rotation=45)
plt.show()

# 2. Bar Chart
plt.figure(figsize=(6,3))
plt.bar(x, df["Entrance Fee in INR"])
plt.title("Bar Chart")
plt.xticks(rotation=45)
plt.show()

# 3. Histogram
plt.figure(figsize=(6,3))
plt.hist(df["Google review rating"], bins=5)
plt.title("Histogram")
plt.show()

# 4. Box Plot
plt.figure(figsize=(4,4))
plt.boxplot(df["Google review rating"].dropna())
plt.title("Box Plot")
plt.show()

# 5. Scatter Plot
plt.figure(figsize=(6,3))
plt.scatter(df["Entrance Fee in INR"],
            df["Google review rating"])
plt.xlabel("Entrance Fee")
plt.ylabel("Rating")
plt.title("Scatter Plot")
plt.show()

# 6. Violin Plot
plt.figure(figsize=(4,4))
sns.violinplot(y=df["Google review rating"])
plt.title("Violin Plot")
plt.show()

# 7. Pie Chart
plt.figure(figsize=(5,5))
plt.pie(df["Entrance Fee in INR"],
        labels=x,
        autopct="%1.1f%%")
plt.title("Pie Chart")
plt.show()

# 8. Subplots (2x2)
fig, ax = plt.subplots(2,2, figsize=(10,8))

ax[0,0].plot(x, df["Google review rating"])
ax[0,0].set_title("Line")

ax[0,1].bar(x, df["Entrance Fee in INR"])
ax[0,1].set_title("Bar")

ax[1,0].hist(df["Google review rating"])
ax[1,0].set_title("Histogram")

ax[1,1].scatter(df["Entrance Fee in INR"],
                df["Google review rating"])
ax[1,1].set_title("Scatter")

for a in ax.flat:
    a.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# 9. Pair Plot
sns.pairplot(df[cols])
plt.show()

# 10. Heatmap
plt.figure(figsize=(5,4))
sns.heatmap(df[cols].corr(),
            annot=True,
            cmap="coolwarm")
plt.title("Heatmap")
plt.show()

# 11. Seaborn 3-Panel Plot
fig, ax = plt.subplots(1,3, figsize=(15,4))

sns.histplot(df["Google review rating"], ax=ax[0])
ax[0].set_title("Histogram")

sns.boxplot(y=df["Google review rating"], ax=ax[1])
ax[1].set_title("Box Plot")

sns.scatterplot(
    x=df["Entrance Fee in INR"],
    y=df["Google review rating"],
    ax=ax[2]
)
ax[2].set_title("Scatter Plot")

plt.tight_layout()
plt.show()
```
